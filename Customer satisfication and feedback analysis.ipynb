```python
# Download stopwords if not already downloaded
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
```

```python
def preprocess_text(text):
    # Remove non-alphabetic characters
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    # Convert to lowercase
    text = text.lower()
    # Remove stopwords
    tokens = text.split()
    tokens = [word for word in tokens if word not in stop_words]
    return ' '.join(tokens)
```

```python
# Apply preprocessing of the reviews
df['processed_reviews'] = df['reviews'].apply(preprocess_text)
```

```python
# Example review
review = "The food was fantastic but the crew was rude."

# Step 1: Aspect keywords
aspects = {
    'food': ['food', 'meal', 'cuisine', 'dish'],
    'crew': ['crew', 'staff', 'service'],
    'fantastic':['fantastic','good','great','delicious']
}

# Step 2: Sentiment analysis
blob = TextBlob(review)
sentiment_score = blob.sentiment.polarity
if sentiment_score > 0.1:
    sentiment = 'positive'
elif sentiment_score < -0.1:
    sentiment = 'negative'
else:
    sentiment = 'neutral'

# Step 3: Aspect detection
for aspect, keywords in aspects.items():
    if any(k in review.lower() for k in keywords):
        print(f"{aspect}: {sentiment}")
```

```python
# To identify where the most positive, negative, or neutral reviews are about specific aspects using words with high frequency from the reviews
data = {
    'output': [
        "The flight was comfortable and the staff were great.",
        "Seat was not very comfortable, but the food was delicious.",
        "Crew was rude and unhelpful during the flight.",
        "British Airways provided excellent service and a smooth flight.",
        "Food quality was poor, but the crew was friendly."
    ]
}
df = pd.DataFrame(data)

# List of frequent words (aspects)
aspects_keywords = {
    'flight': ['flight', 'fly', 'airline', 'airlines'],
    'service': ['service', 'staff', 'crew', 'attentive', 'rude', 'friendly'],
    'seat': ['seat', 'seats', 'comfort', 'comfortable'],
    'food': ['food', 'meal', 'dinner', 'lunch', 'breakfast', 'quality'],
    'british': ['british', 'london', 'uk', 'uk-based']
}

# Function to get overall sentiment of review
def get_sentiment(text):
    blob = TextBlob(text)
    polarity = blob.sentiment.polarity
    if polarity > 0.1:
        return 'positive'
    elif polarity < -0.1:
        return 'negative'
    else:
        return 'neutral'

# Function to extract aspects and sentiment
def extract_aspects_sentiment(review_text):
    review_lower = review_text.lower()
    review_sentiment = get_sentiment(review_text)
    aspect_sentiments = []

    for aspect, keywords in aspects_keywords.items():
        if any(keyword in review_lower for keyword in keywords):
            aspect_sentiments.append((aspect, review_sentiment))
    return aspect_sentiments

# Apply extraction to dataset
df['aspects_sentiments'] = df['output'].apply(extract_aspects_sentiment)

# Summarize results
from collections import Counter

# Count positive/negative mentions per aspect
aspect_counts = {}
for index, row in df.iterrows():
    for aspect, sentiment in row['aspects_sentiments']:
        if aspect not in aspect_counts:
            aspect_counts[aspect] = {'positive':0, 'negative':0, 'neutral':0}
        aspect_counts[aspect][sentiment] += 1

# Display counts
for aspect, sentiments in aspect_counts.items():
    print(f"\nAspect: {aspect}")
    print(f"Positive mentions: {sentiments['positive']}")
    print(f"Negative mentions: {sentiments['negative']}")
    print(f"Neutral mentions: {sentiments['neutral']}")
```

```python
# Topic modeling with LDA
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
data = {
    'output': [
        "The flight was comfortable and the staff were great.",
        "Seat was not very comfortable, but the food was delicious.",
        "Crew was rude and unhelpful during the flight.",
        "British Airways provided excellent service and a smooth flight.",
        "Food quality was poor, but the crew was friendly."
    ]
}
df = pd.DataFrame(data)

# Vectorize text data
vectorizer = CountVectorizer(stop_words='english', max_df=0.95, min_df=2)
X = vectorizer.fit_transform(df['output'])

# Define number of topics
n_topics = 4

# Fit LDA model
lda = LatentDirichletAllocation(n_components=n_topics, random_state=42)
lda.fit(X)

# Function to display top words in each topic
def display_topics(model, feature_names, n_top_words):
    for topic_idx, topic in enumerate(model.components_):
        print(f"\nTopic #{topic_idx + 1}:")
        top_features_indices = topic.argsort()[:-n_top_words - 1:-1]
        top_features = [feature_names[i] for i in top_features_indices]
        print(" | ".join(top_features))

# Display top words per topic
n_top_words = 5
feature_names = vectorizer.get_feature_names_out()
display_topics(lda, feature_names, n_top_words)
```
